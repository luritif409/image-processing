```python
import cv2

# baca gambar
image = cv2.imread('images.jpg')

# konversi ke grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# simpan hasil
cv2.imwrite('output_grayscale.jpg', gray)
```

```python
import cv2

image = cv2.imread('output_grayscale.jpg', 0)

# thresholding
_, binary = cv2.threshold(image, 128, 255, cv2.THRESH_BINARY)

cv2.imwrite('output_binary.jpg', binary)
```

```python
import cv2
import numpy as np

image = cv2.imread('output_grayscale.jpg', 0)

m = 5  # 3-bit → 8 level
levels = 2 ** m
step = 256 // levels

quantized = (image // step) * step
quantized = quantized.astype(np.uint8)

cv2.imwrite('output_mbit.jpg', quantized)
```

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('output_grayscale.jpg', 0)

plt.figure(figsize=(10,4))

# gambar
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

# histogram
plt.subplot(1,2,2)
plt.hist(image.ravel(), bins=256, range=[0,256])
plt.title('Histogram')

plt.show()
```
