```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np


# membaca gambar
img = cv2.imread('/content/images_marka.jpg')

# membuat negative
negative = 255 - img

# simpan hasil
cv2.imwrite('negative.jpg', negative)

# tampilkan
cv2_imshow(img)
cv2_imshow(negative)

#------------logaritmik------------------
## membaca gambar grayscale
# img = cv2.imread('image.jpg', 0)
# img = cv2.imread('/content/images_marka.jpg')

# ubah ke float agar log bisa dihitung
img_float = np.float32(img)

# konstanta
c = 255 / np.log(1 + np.max(img_float))

# transformasi log
log_image = c * np.log(1 + img_float)

# normalisasi ke uint8
log_image = np.array(log_image, dtype=np.uint8)

# simpan hasil
cv2.imwrite('log_transform.jpg', log_image)

# tampilkan
# cv2_imshow(img)
cv2_imshow(log_image)


#----------Power Law-------------------
# normalisasi ke range 0-1
img_norm = img / 255.0

# nilai gamma
gamma = 2.0

# konstanta
c = 1

# transformasi power-law
power_image = c * np.power(img_norm, gamma)

# kembalikan ke range 0-255
power_image = np.uint8(power_image * 255)

# simpan hasil
cv2.imwrite('power_law.jpg', power_image)

# tampilkan
# cv2_imshow(img)
cv2_imshow(power_image)
```
