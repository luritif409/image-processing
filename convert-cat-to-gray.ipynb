```python

import cv2
import matplotlib.pyplot as plt


img = cv2.imread('/content/kucing.jpg')
img = cv2.imread('/content/kucing.jpg',cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
# plt.imshow(img)
plt.title('Citra Asli (Grayscale)')
plt.show()
```
